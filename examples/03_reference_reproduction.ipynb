# Frozen reference reproduction
Use versioned rankings and assignments when reproducing a manuscript dataset.

```python
import scanpy as sc
from sigma_spatial import get_reference_preset, run_reference_analysis

get_reference_preset("HBC515")
```

```python
adata = sc.read_h5ad("HBC515_SIGMA_core.h5ad")
report = run_reference_analysis(
    adata, dataset="HBC515",
    ranking_path="BC515_lambda_ranking.csv",
    assignments_path="BC515_program_assignments.csv",
    output_dir="results/HBC515_reference",
)
report.leading_program
```
