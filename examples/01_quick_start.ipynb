# SIGMA quick start
Load one prepared AnnData object, inspect its evidence, and generate a standard report.

```python
import scanpy as sc
from sigma_spatial import inspect_input, run_analysis

adata = sc.read_h5ad("sample.h5ad")
inspect_input(adata).print_report()
```

```python
result = run_analysis(
    adata, output_dir="results/sample", prefix="sample",
    workflow="auto", evidence={"same_section_pathology": True},
    report_level="complete", random_state=0,
)
result.downstream.leading_program
```
